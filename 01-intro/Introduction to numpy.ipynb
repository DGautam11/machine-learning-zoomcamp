```python
import matplotlib.pyplot as plt
```

```python
img = plt.imread('dc_metro.png')
plt.imshow(img, cmap=plt.cm.hot)
plt.show()
```

```python

def blur(img, times = 50):
    
    if times <= 0:
        return img
        
    center = img[1:-1,1:-1]
    top = img[:-2,1:-1]
    left = img[1:-1,:-2]
    bottom= img[2:,1:-1]
    right= img[1:-1,2:]

    blurred = (top + left + right + bottom + center) / 5

    return blur(blurred, times-1)

blurred_img = blur(img)
plt.figure()
plt.imshow(blurred_img,cmap=plt.cm.hot)
plt.show()
```
